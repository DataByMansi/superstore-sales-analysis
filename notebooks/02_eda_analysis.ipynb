# NOTEBOOK 2: EXPLORATORY DATA ANALYSIS (EDA)

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('../data/superstore_clean.csv')
```

```python
# --- 1. Summary Statistics ---
print("=== SUMMARY STATISTICS ===")
print(df[['Sales', 'Profit', 'Discount', 'Quantity']].describe())
```

```python
# --- 2. Sales by Category ---
category_sales = df.groupby('Category')[['Sales', 'Profit']].sum().reset_index()
category_sales['Profit_Margin_%'] = (category_sales['Profit'] / category_sales['Sales'] * 100).round(2)
print("\n=== SALES BY CATEGORY ===")
print(category_sales)
```

```python
# --- 3. Sales by Region ---
region_sales = df.groupby('Region')[['Sales', 'Profit']].sum().reset_index()
print("\n=== SALES BY REGION ===")
print(region_sales)
```

```python
# --- 4. Top 10 Products by Sales ---
top_products = df.groupby('Product_Name')['Sales'].sum().nlargest(10).reset_index()
print("\n=== TOP 10 PRODUCTS ===")
print(top_products)
```

```python
# --- 5. Loss-Making Products ---
loss_products = df.groupby('Product_Name')['Profit'].sum().nsmallest(10).reset_index()
print("\n=== TOP 10 LOSS-MAKING PRODUCTS ===")
print(loss_products)
```

```python
# --- 6. Yearly Sales Trend ---
yearly = df.groupby('Order_Year')[['Sales', 'Profit']].sum().reset_index()
print("\n=== YEARLY TREND ===")
print(yearly)
```

```python
# --- 7. Discount Impact on Profit ---
discount_impact = df.groupby('Discount')[['Profit', 'Sales']].mean().reset_index()
print("\n=== DISCOUNT vs PROFIT ===")
print(discount_impact)
```

```python
# --- 8. Ship Mode Analysis ---
ship_mode = df.groupby('Ship_Mode')[['Sales', 'Days_to_Ship']].mean().reset_index()
print("\n=== SHIP MODE ANALYSIS ===")
print(ship_mode)
```

```python
# Export summary tables
category_sales.to_csv('../exports/category_sales.csv', index=False)
region_sales.to_csv('../exports/region_sales.csv', index=False)
```
